```python

from google.colab import drive

drive.mount('/content/drive')
import pandas as pd
DATA_PATH = "/content/drive/MyDrive/titanic.csv"
df = pd.read_csv(
    DATA_PATH
)
df.head()
```

```python
df.drop(['PassengerId', 'Name', 'SibSp', 'Parch', 'Ticket', 'Cabin', 'Embarked'], axis='columns', inplace=True)
df.head()
```

```python
target = df[["Survived"]]
target.head()
```

```python
input = df.drop('Survived',axis='columns')
input.head()
```

```python
dummies = pd.get_dummies(input.Sex)
dummies.head()
```

```python
input = pd.concat([input, dummies], axis='columns')
input.head()
```

```python
input = input.drop('Sex',axis='columns')
input
```

```python
input.columns[input.isna().any()]
input.Age[:10]
```

```python
input.Age = input.Age.fillna(input.Age.mean())
input.Age[:10]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(input,target)
```

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
model.fit(x_train,y_train)
```

```python
model.score(x_train,y_train)
```

```python
from sklearn.metrics import classification_report,confusion_matrix,accuracy_score
y_predicted = model.predict(x_test)
cm = confusion_matrix(y_test,y_predicted)
cm
```

```python
print("Classification report: ",classification_report(y_test,y_predicted))
```
