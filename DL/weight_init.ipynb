```python
from tensorflow.keras.initializers import HeNormal,GlorotNormal
from tensorflow.keras.layers import Input,Dense,Flatten,Dropout
from tensorflow.keras.models import Sequential
from tensorflow.keras import regularizers
from tensorflow.keras.datasets import cifar10
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
(x_train,y_train),(x_test,y_test) = cifar10.load_data()
x_train,x_test = x_train/255.0,x_test/255.0
```

```python
base_model = Sequential([
    Input(shape=(32,32,3)),
    Flatten(),
    Dense(256,activation='relu'),
    Dense(128,activation='relu'),
    Dense(10,activation='softmax')
])
```

```python
xavier_model = Sequential([
    Input(shape=(32,32,3)),
    Flatten(),
    Dense(256,activation='relu',kernel_initializer=GlorotNormal()),
    Dense(128,activation='relu',kernel_initializer=GlorotNormal()),
    Dense(10,activation='softmax')
])
```

```python
kaiming_model = Sequential([
    Input(shape=(32,32,3)),
    Flatten(),
    Dense(256,activation='relu',kernel_initializer=HeNormal()),
    Dense(128,activation='relu',kernel_initializer=HeNormal()),
    Dense(10,activation='softmax')
])
```

```python
dropout_model = Sequential([
    Input(shape=(32,32,3)),
    Flatten(),
    Dense(256,activation='relu'),
    Dropout(0.5),
    Dense(128,activation='relu'),
    Dropout(0.5),
    Dense(10,activation='softmax')
])
```

```python
model_l1 = Sequential([
    Input(shape=(32,32,3)),
    Flatten(),
    Dense(256,activation='relu',kernel_regularizer=regularizers.L1()),
    Dense(128,activation='relu',kernel_regularizer=regularizers.L1()),
    Dense(10,activation='softmax')
])
```

```python
model_l2 = Sequential([
    Input(shape=(32,32,3)),
    Flatten(),
    Dense(256,activation='relu',kernel_regularizer=regularizers.L2()),
    Dense(128,activation='relu',kernel_regularizer=regularizers.L2()),
    Dense(10,activation='softmax')
])
```

```python
base_model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
xavier_model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
kaiming_model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
dropout_model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model_l1.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model_l2.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
base_history = base_model.fit(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```

```python
xavier_history = xavier_model.fit(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```

```python
kaiming_history = kaiming_model.fit(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```

```python
dropout_history = dropout_model.fit(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```

```python
l1_history = model_l1.fit(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```

```python
l2_history = model_l2.fit(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```

```python
base_pred=base_model.predict(x_test)
```

```python
xavier_pred=xavier_model.predict(x_test)
```

```python
kaiming_pred=kaiming_model.predict(x_test)
```

```python
dropout_pred=dropout_model.predict(x_test)
```
